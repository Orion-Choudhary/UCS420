## Q.6 Create a sample dataset (employees.csv) containing information about employees in a company.




```python

employees = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
})

employees.to_csv("employees.csv", index=False)
employees
```

### e) Calculate the following statistics from the dataset:

#### i. The average salary of employees.


```python
print("Average salary of employees is: ", employees["Salary"].mean())
```

#### ii. The total bonus paid to all employees.


```python
print("Total bonus of all employees are: ", employees["Bonus"].sum())
```

#### iii. The youngest employee's age.


```python
print("Age of the youngest employee is: ", employees["Age"].min())
```

#### iv. The highest performance rating.


```python
print("The highest performance rating among all employees is: ", employees["Rating"].max())
```

### f) Sort the DataFrame by the Salary column in descending order.


```python
employees = employees.sort_values(by= "Salary", ascending = False)
print(employees)
```

### g) Add a new column that categorizes employees based on their performance rating:
#### i. Excellent for ratings >= 4.5
#### ii. Good for ratings >= 4.0 but < 4.5
#### iii. Average for ratings < 4.0


```python
def perf_category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

employees["Performance"] = employees["Rating"].apply(perf_category)

print(employees)
```

### h) Identify missing values in the DataFrame.


```python
print(employees.isnull().sum())
```

### i) Rename the Employee_ID column to ID.


```python
employees.rename(columns={"Employee_ID":"ID"}, inplace = True)
print("Column is successfully renamed: ")
print(employees)
```

### j) Find all employees who:
#### i. Have more than 5 years of experience.


```python
print("Employees with experience greater than 5 years: ")
print(employees[employees["Years_of_Experience"]>5])
```

#### ii. Belong to the IT department.


```python
print("Employees who belong to the IT Department: ")
print(employees[employees["Department"]=="IT"])
```

### k) Modify the dataset by adding a new column, Tax, which deducts 10% of the Salary.


```python
employees["Tax"] = employees["Salary"]* 0.10
print("New Column Successfullly added.")
print(employees)
```
