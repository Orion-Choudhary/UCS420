### a-d. Basic Slicing, Views vs Copies (`np.copy()`)

````python
original = np.array([1, 2, 3, 4, 5, 6])

subset = original[1:5]
subset[0] = 999
print("Original:", original) 
print("Subset:", subset)

original_2 = np.array([1, 2, 3, 4, 5, 6])
subset_copy = original_2[1:5].copy()
subset_copy[0] = 500
print("Original:", original_2)  
print("Copied Subset:", subset_copy)

````

## Q4. Cognitive Assistive System using Ordinary Least Squares (OLS)

````python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])
````

### a-d. Matrix Attributes, Transpose, & Matrix Inversion

````python
print("Shape of X is:", X.shape, " and dimensions are:", X.ndim)

X_T = X.T
print("The shape of X.T is:", X_T.shape)
print("The transpose swaps rows and columns. This converts features into rows and each user into its own separate column.")


XTX = X_T @ X
print("X.T @ X:\n", XTX)

XTX_inverse = np.linalg.inv(XTX)
print("Inverse is equivalent to:\n", XTX_inverse)

````

### e-f. Solve $\beta = (X^T X)^{-1} X^T y$ and interpretation

````python
beta = XTX_inverse @ X_T @ y
print("The Coefficients are equal to:", beta)

print("Coefficient Interpretation:")
print(f"  - Sleep Hours (x1 weight): {beta[0]:.2f} > Each extra hour of sleep changes predicted assistance score by {beta[0]:.2f}.")
print(f"  - Activity Level (x2 weight): {beta[1]:.2f} > Each unit of activity level changes assistance score by {beta[1]:.2f}.")
print(f"  - Stress Level (x3 weight): {beta[2]:.2f} > Each unit of stress level changes assistance score by {beta[2]:.2f}.")

````

### g. Prediction for new user (`new_user = np.array([5, 40, 7])`)

````python
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print(f"Predicted assistance score for new user {new_user}: {predicted_score:.2f}")
````
