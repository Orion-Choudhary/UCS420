## Q.2 Questions on Basic NumPy Array

### a) Reverse the NumPy array: `arr = np.array([1, 2, 3, 6, 4, 5])`

```python
arr = np.array([1, 2, 3, 6, 4, 5])

rev_arr= arr[::-1]
print("The reversed array: ",rev_arr)

```

### b) Find the most frequent value and their indice(s) in the following arrays:

#### i. `x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])`

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
vals_x, counts_x = np.unique(x, return_counts=True)
most_freq_x = vals_x[np.argmax(counts_x)]
indices_x = np.where(x == most_freq_x)[0]
print(f" Most frequent value in the provided array x is: {most_freq_x}, located at the Indices: {indices_x}")
```

#### ii. `y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])`

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
vals_y, counts_y = np.unique(y, return_counts=True)
max_count_y = np.max(counts_y)
most_freq_y = vals_y[counts_y == max_count_y] 
print(f"Most frequent value(s) present in the provided array y is: {most_freq_y}")
for val in most_freq_y:
    indices_y = np.where(y == val)[0]
    print(f" Indices for the required value {val}: {indices_y}")
```

## Q.4 Array Creation with `linspace()` and Attributes Analysis

1. Create a 1D NumPy array named `<Your Name>` with 25 evenly spaced numbers from 10 to 100 using `linspace()`.
2. Print dimensions, shape, total elements, data type of elements, and total bytes consumed.
3. Find the transpose using `reshape()` attribute.
4. Check if the same transpose can be performed using the `.T` attribute.

```python
Utkarsh_Choudhary = np.linspace(10, 100, 25)

print("Array with linspace: ", Utkarsh_Choudhary)
print("Dimensions of the array:", Utkarsh_Choudhary.ndim)
print("Shape of the array:", Utkarsh_Choudhary.shape)
print("Total elements:", Utkarsh_Choudhary.size)
print("Data type of elements present in the array:", Utkarsh_Choudhary.dtype)
print("Total bytes consumed:", Utkarsh_Choudhary.nbytes)

t_reshape = Utkarsh_Choudhary.reshape(-1, 1)
print("Transposed using reshape() shape:", t_reshape.shape)

transposed_T = Utkarsh_Choudhary.T
print("Transposed using .T shape:", transposed_T.shape)
```

## Q.5 2-D Array Manipulations, Statistics, Reshaping, and Resizing

1. Create a 2D array of 3 rows and 4 columns named `ucs420_<your_name>` with values: `10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35`.
2. Compute mean, median, max, min, and unique elements.
3. Reshape array to 4 rows and 3 columns (`reshaped_ucs420_<your_name>`).
4. Resize array to 2 rows and 3 columns (`resized_ucs420_<your_name>`).

```python
data = [10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35]

ucs420_UtkarshChoudhary = np.array(data).reshape(3, 4)
print("The formed 3x4 Array:", ucs420_UtkarshChoudhary)

print("Mean:", np.mean(ucs420_UtkarshChoudhary))
print("Median:", np.median(ucs420_UtkarshChoudhary))
print("Max:", np.max(ucs420_UtkarshChoudhary))
print("Min:", np.min(ucs420_UtkarshChoudhary))
print("Unique elements:", np.unique(ucs420_UtkarshChoudhary))

reshaped_ucs420_UtkarshChoudhary = ucs420_UtkarshChoudhary.reshape(4, 3)
print("Reshaped Array (4x3):\n", reshaped_ucs420_UtkarshChoudhary)

resized_ucs420_UtkarshChoudhary = np.resize(ucs420_UtkarshChoudhary, (2, 3))
print("Resized Array (2x3):\n", resized_ucs420_UtkarshChoudhary)
```
